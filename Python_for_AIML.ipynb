```python
#file handling project
from pathlib import Path
import os
def readfileandfolder():
  path = Path('')
  items = list(path.rglob('*'))
  for i , items in enumerate(items):
     print(f"{i+1} : {items}")


def create_file():
  try:
   readfileandfolder()
   name = input ("Please tell your file name  : ")
   p = Path(name)
   if not p.exists():
    with open(p, "w") as fs:
      data = input("what you want to write in this file :-")
      fs.write(data)
      print(f"FILE CREATED SUCCESSFULLY")
   else:
        print("This file already exists")
  except Exception as err:
    print(f"An error occured as {err}")



def read_file():
  try:
    readfileandfolder()
    name = input("Please enter your file name")
    path = Path(name)
    if path.exists and path.is_file():
      with open(path , "r") as fs:
        data = fs.read()
        print(data)
      print("Readed Successfully")
    else:
      print("The file doesn't exist")
  except Exception as err:
    print(f"An error occured as {err}")


def update_file():
    try:
      readfileandfolder()
      name = input("Please tell your file name you want to update:")
      p = Path(name)
      if p.exists() and p.is_file():
        print("Press 1 for changing the file name: ")
        print("Press 2 for overwriting the data of your file : ")
        print("Press 3 for appending the file data: ")

        res = int(input("Enter your choice:- "))

        if res == 1:
          name2 = input("Tell your new file name:- ")
          p2 = Path(name2)
          p.rename(p2)

        if res == 2:
          with open(p, "w") as fs:
           data = input("Enter the text to overwrite : ")
           fs.write(data)

        if res == 3:
          with open(p, "w") as fs:
           data = input("Enter the text to overwrite : ")
           fs.write(" " + data)

    except Exception as err:
      print(f"An error has occured as {err}")


def delete_file():
  try:
    readfileandfolder()
    name = input("Enter file name to be deleted:- ")
    p = Path(name)

    if p.exists() and p.is_file():
      os.remove(name)
      print("FILE REMOVED SUCCESSFULLY")
    else:
      print("No such file exist")


  except Exception as err:
    print(f"An error occured as {err}")



print("Press 1 for creating a file")
print("Press 2 for Reading a file")
print("Press 3 for Updating a file")
print("Press 4 for delete a file")
check = int(input("Enter your choice : "))
if check == 1:
  create_file()
elif check == 2:
  read_file()
elif check == 3:
  update_file()
elif check == 4:
  delete_file()
else :
  print("Invalid Input")
```

```python
from abc import ABC, abstractmethod

# Abstract Class User
class User(ABC):
    def __init__(self, user_id, name, email, password):
        self.user_id = user_id
        self.name = name
        self.email = email
        # Encapsulate password
        self.__password = password

    def login(self, entered_password):
        if self.__password == entered_password:
            print(f"User {self.name} logged in successfully.")
            return True
        else:
            print("Invalid password.")
            return False

    def logout(self):
        print(f"User {self.name} logged out.")

    @abstractmethod
    def display_profile(self):
        pass

# Derived Class Student
class Student(User):
    def __init__(self, user_id, name, email, password, roll_number):
        super().__init__(user_id, name, email, password)
        self.roll_number = roll_number
        self.enrolled_courses = []  # List of Course objects
        self.exam_history = []      # List of Result objects
        self.grades = {}            # Dictionary: course_id -> grade

    def display_profile(self):
        print(f"\n--- Student Profile ---")
        print(f"User ID: {self.user_id}")
        print(f"Name: {self.name}")
        print(f"Email: {self.email}")
        print(f"Roll Number: {self.roll_number}")
        print(f"Enrolled Courses: {[course.course_name for course in self.enrolled_courses]}")
        print(f"Grades: {self.grades}")

    def enroll_course(self, course):
        self.enrolled_courses.append(course)
        print(f"Student {self.name} enrolled in {course.course_name}.")

    def take_exam(self, exam):
        # Placeholder for taking an exam
        print(f"Student {self.name} is taking exam {exam.exam_id} for {exam.course_id}.")

    def view_results(self):
        # Placeholder for viewing results
        print(f"Displaying results for {self.name}.")

    def download_material(self, material):
        # Placeholder for downloading material
        print(f"Student {self.name} downloading {material.title}.")

# Derived Class Teacher
class Teacher(User):
    def __init__(self, user_id, name, email, password, employee_id, specialization):
        super().__init__(user_id, name, email, password)
        self.employee_id = employee_id
        self.specialization = specialization
        self.assigned_courses = []  # List of Course objects

    def display_profile(self):
        print(f"\n--- Teacher Profile ---")
        print(f"User ID: {self.user_id}")
        print(f"Name: {self.name}")
        print(f"Email: {self.email}")
        print(f"Employee ID: {self.employee_id}")
        print(f"Specialization: {self.specialization}")
        print(f"Assigned Courses: {[course.course_name for course in self.assigned_courses]}")

    def create_course(self, course_name, course_id):
        # Placeholder for creating a course
        print(f"Teacher {self.name} created course {course_name}.")

    def upload_material(self, course, material):
        # Placeholder for uploading material
        print(f"Teacher {self.name} uploaded material {material.title} for {course.course_name}.")

    def create_exam(self, course_id, exam_id):
        # Placeholder for creating an exam
        print(f"Teacher {self.name} created exam {exam_id} for course {course_id}.")

    def manage_question_bank(self, question_bank):
        # Placeholder for managing question bank
        print(f"Teacher {self.name} managing question bank {question_bank.subject}.")


# Example Usage (for testing):
student1 = Student("S001", "Alice", "alice@example.com", "pass123", "R001")
teacher1 = Teacher("T001", "Mr. Smith", "smith@example.com", "teachpass", "E001", "Mathematics")

student1.login("pass123")
student1.display_profile()

teacher1.login("wrongpass")
teacher1.display_profile()
```

# boxplots and violinplots
these are used to shown the distribution of categorical data.

A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be "outliers" using a method that is a function of the inter-quartile range.

```python
sns.violinplot(x = df['day'], y = 'total_bill', data = df)
```

```python
sns.violinplot(y = df['day'], x = 'total_bill', data = df)
```

# strip plot and swarm plot

The strip plot will draw a scatter plot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarm plot is similar to strip plot(), but the points are adjusted (only along the categorical axis) so that they don’t overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arrange them).

```python
sns.stripplot(x = 'day', y = "total_bill",  data = df )
```

```python
sns.swarmplot(x = 'day', y = "total_bill",  data = df )
```

```python
sns.violinplot(x = "tip", y = "day", data = df)
sns.swarmplot(x = "tip", y = "day", data = df)
```
